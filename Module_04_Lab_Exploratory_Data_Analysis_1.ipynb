## Part 1: Setup and Data Loading**What is Exploratory Data Analysis (EDA)?**EDA is the process of using summary statistics and visualizations to understand a dataset's main characteristics. Before you can build a model, you need to understand your data. What stories does it tell? Are there errors or missing values? Are there strong relationships between variables? EDA helps answer these questions.We will use the famous Titanic dataset for this lab. It contains information about passengers and, crucially, whether they survived the disaster.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

print("--- First 5 Rows ---")
print(df.head())

print("\n--- Basic Info ---")
df.info()
```

## Part 2: Descriptive StatisticsLet's start by getting a high-level numerical summary of the data. The `.describe()` method is perfect for this. It calculates statistics like mean, standard deviation, min, and max for the numerical columns.

```python
print("--- Descriptive Statistics ---")
print(df.describe())

print("\n--- Key Insights from Statistics ---")
print(f"The average age of a passenger was {df['Age'].mean():.1f} years.")
print(f"The overall survival rate was {df['Survived'].mean():.1%}.")
print(f"Fares ranged from ${df['Fare'].min()} to a whopping ${df['Fare'].max()}.")
```

## Part 3: Visual EDA - Telling Stories with PlotsNumbers are great, but plots make patterns and relationships immediately obvious. The goal of visual EDA is to turn data into insights.**A Note on Plotting Libraries:***   **Matplotlib:** The foundational library, gives you full control over everything.*   **Seaborn:** Built on Matplotlib, it provides a simpler, high-level interface for creating common statistical plots. We will use Seaborn for its ease of use and attractive defaults.

### Visualization 1: How many survived?A simple `countplot` is the best way to see the distribution of a categorical variable, like our target `Survived`.

```python
sns.set_style('whitegrid')
plt.figure(figsize=(8, 6))
sns.countplot(x='Survived', data=df)
plt.title('Survival Distribution (0 = Died, 1 = Survived)')
plt.show()

print("Insight: Far more people died than survived. This is an example of an imbalanced dataset, which can sometimes be a challenge for machine learning models.")
```

### Visualization 2: Does passenger class matter for survival?Now we want to see if there's a relationship between two variables. We can use the `hue` parameter in `countplot` to split the bars by another category.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival by Passenger Class')
plt.legend(['Died', 'Survived'])
plt.show()

print("Insight: This is a very strong pattern. 1st class passengers had a much higher chance of survival compared to 3rd class passengers. Money seems to have made a difference.")
```

### Visualization 3: What about gender?The 'women and children first' mantra is famous. Let's see if the data supports it.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.legend(['Died', 'Survived'])
plt.show()

print("Insight: The pattern is undeniable. A much higher proportion of females survived compared to males. This is another very strong predictor.")
```

### Visualization 4: How does age play a role?For a continuous variable like `Age`, a histogram is a great way to see its distribution.

```python
g = sns.FacetGrid(df, col='Survived', height=6)
g.map(plt.hist, 'Age', bins=20)
plt.show()

print("Insight: The age distribution for those who did not survive is centered around the 20-40 age range. For those who survived, there is a noticeable spike for young children. This supports the 'children' part of the mantra.")
```

## Part 4: Student Experimentation**Instructions:** Create your own visualizations to explore other relationships in the data.

### Experiment 1: Port of Embarkation1.  The `Embarked` column tells you where a passenger boarded the ship (C = Cherbourg, Q = Queenstown, S = Southampton).2.  Create a `countplot` to see how survival rates differed by the port of embarkation. Does where they boarded seem to be related to their survival?

```python
plt.figure(figsize=(9, 6))
sns.countplot(x="Embarked", hue="Survived", data=df, palette="Set2")
plt.title("Survival by Port of Embarkation (C=Cherbourg, Q=Queenstown, S=Southampton)")
plt.xlabel("Port of Embarkation")
plt.ylabel("Passenger Count")
plt.legend(["Died", "Survived"])
plt.show()

print("--- Survival Proportion by Port ---")
print(df.groupby("Embarked")["Survived"].value_counts(normalize=True).unstack())
print("\nInsight: Cherbourg (C) had a higher survival rate (>50%) compared to Southampton (S) and Queenstown (Q).")
print("This correlates strongly with socioeconomic status, as Cherbourg had a significantly higher proportion of 1st-class ticket holders.")
```

### Experiment 2: Fare vs. Survival1.  `Fare` is a continuous numerical variable.2.  A `boxplot` or `violinplot` is a great way to see the distribution of fares for those who survived vs. those who didn't.3.  Create one of these plots to compare the `Fare` distribution by `Survived`. What does it tell you?

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Boxplot full range
sns.boxplot(ax=axes[0], x="Survived", y="Fare", data=df, palette="Blues")
axes[0].set_title("Fare Distribution by Survival Status (Full Range)")
axes[0].set_xticklabels(["Died", "Survived"])
axes[0].set_ylabel("Fare ($)")

# Zoomed view under extreme outliers
sns.boxplot(ax=axes[1], x="Survived", y="Fare", data=df, palette="Blues")
axes[1].set_ylim(0, 150)
axes[1].set_title("Fare Distribution by Survival Status (Zoomed: $0-$150)")
axes[1].set_xticklabels(["Died", "Survived"])
axes[1].set_ylabel("Fare ($)")

plt.tight_layout()
plt.show()

print("Median Fare (Died):    $", df[df["Survived"] == 0]["Fare"].median())
print("Median Fare (Survived):$", df[df["Survived"] == 1]["Fare"].median())
print("\nInsight: Passengers who survived paid a significantly higher median fare ($26.00) than those who perished ($10.50).")
print("The fare distribution is heavily right-skewed, showing extreme high-fare outliers exclusively surviving or receiving lifeboat priority.")
```

**1. What is the primary goal of Exploratory Data Analysis (EDA)?**
The primary goal of EDA is to uncover underlying structures, identify patterns and anomalies, audit data quality (such as missing values and skewed distributions), and evaluate relationships between variables prior to modeling. It provides the empirical diagnostic context needed to choose correct preprocessing techniques, handling strategies for missing data, and feature representations before training machine learning algorithms.

**2. Based on the plots in this lab, what kind of person had the best chance of surviving the Titanic?**
A **female child or woman traveling in 1st Class** had the highest statistical likelihood of survival:
* **Gender:** Over 70% of females survived compared to under 20% of males, making sex the single most decisive categorical predictor.
* **Class:** 1st Class passengers experienced the highest survival proportion, whereas 3rd Class passengers suffered the vast majority of casualties.

**3. Why is it important to visualize data instead of just looking at summary statistics? What can a plot show you that a number like 'mean' or 'count' can't?**
Summary statistics compress multi-dimensional distributions into single numerical values, which often hide critical data behaviors:
* **Skewness and Extreme Outliers:** The mean fare of approximately $32.20 is artificially pulled upward by extreme outlier tickets exceeding $500. A boxplot clearly separates the typical passenger reality (median around $14) from high-leverage outliers.
* **Multimodal Patterns:** A single mean age of ~29.7 years conceals whether the population is uniform, bell-shaped, or split into clusters. A histogram or FacetGrid immediately exposes the disproportionate survival spike among young children.
* **Complex Subgroup Relationships:** Summary metrics cannot easily demonstrate non-linear interactions across multiple dimensions, whereas split plots (like countplots with `hue` or bivariate grids) display category-level disparities immediately.**
